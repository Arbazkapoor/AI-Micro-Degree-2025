```python
# Step-1: Import Necessary Libraries
import cv2
import os
```

```python
# Step-2: Load the pre-trained Haar cascade for face detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml') 
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_eye.xml') 
```

```python
# Step-3: Start the webcam
cap = cv2.VideoCapture(0)
```

```python
# Step-4: Check if webcam opened successfully
if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()
```

```python
# Step 5: Keep reading images (frames) from the webcam
while True:
    ret, frame = cap.read()  # Take a photo (frame) from the webcam
    if not ret:  # If the camera didn’t give a frame, stop the loop
        break

    # Change the photo to black and white (gray), because face detection works better in gray
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Find faces in the gray image
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

    # For every face found, draw a blue rectangle around it
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)

        # Now check the area inside the face to find eyes
        roi_gray = gray[y:y + h, x:x + w]      # Gray part inside the face
        roi_color = frame[y:y + h, x:x + w]    # Color part inside the face

        # Find eyes in that gray area (inside the face)
        eyes = eye_cascade.detectMultiScale(roi_gray, scaleFactor=1.1, minNeighbors=4)

        # For every eye found, draw a green rectangle around it
        for (ex, ey, ew, eh) in eyes:
            cv2.rectangle(roi_color, (ex, ey), (ex + ew, ey + eh), (0, 255, 0), 2)

    # Show the video with the rectangles drawn on it
    cv2.imshow('Face Detection - Webcam', frame)

    # If the user presses the letter 'q', exit the loop
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Step 6: Close the camera and window properly
cap.release()  # Turn off the webcam
cv2.destroyAllWindows()  # Close the window where the video was shown
```
